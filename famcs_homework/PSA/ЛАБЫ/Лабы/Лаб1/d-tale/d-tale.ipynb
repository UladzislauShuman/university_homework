```python
import pandas as pd
import dtale

try:
    df = pd.read_csv('../data.csv', header=None, names=['Период', 'Значение_xi'], decimal=',')
    df = df.sort_values(by='Значение_xi').reset_index(drop=True)
except:
    print("Ошибка: Проверь файл data.csv")



df['Частота_mi'] = 1

total_count = len(df)
df['Частость_wi'] = df['Частота_mi'] / total_count

df['Накопл_частота_mi_C'] = df['Частота_mi'].cumsum()

df['Накопл_частость_wi_C'] = df['Частость_wi'].cumsum()

d = dtale.show(df)
d.open_browser()
```
