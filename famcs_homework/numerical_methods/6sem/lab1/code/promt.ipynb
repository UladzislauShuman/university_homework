    • ось X — номер итерации k
    • ось Y — ошибка x_k - x*
      или невязка f(x_k)
    • логарифмическая шкала по основанию 10

Для задачи 1.11:
    - Построить совмещенные диаграммы сходимости:
        • метода обратной квадратичной интерполяции (ОКИ)
        • метода секущих
        • метода Ньютона
    - Подтвердить порядок сходимости экспериментально.

Для задачи 2.9:
    - Определить количество вещественных корней системы.
    - Найти все корни методом Ньютона.
    - Построить диаграммы сходимости для каждого корня.

ЭТАП 6. Код
- Код должен:
    • точно реализовывать описанный алгоритм
    • быть аккуратно оформленным
    • не содержать лишних частей
    • соответствовать математическому описанию
- Автор должен понимать каждую строку.
